```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from scipy.stats import pearsonr
```

```python
from google.colab import files
uploaded = files.upload()

df1 = pd.read_csv('grupo1.csv')
df2 = pd.read_csv('grupo2.csv')
df3 = pd.read_csv('grupo3.csv')
```

```python
print('GRUPO 1')
print(df1.head())
print(df1.describe())
plt.figure(figsize=(6,4))
plt.hist(df1['Calificacion'], bins=10)
plt.title('Calificaciones del Grupo 1')
plt.xlabel('Calificación')
plt.ylabel('Frecuencia')
plt.grid(True)
plt.show()
```

```python
print('GRUPO 2')
print(df2.head())
print(df2.describe())
plt.figure(figsize=(6,4))
plt.hist(df2['Calificacion'], bins=10)
plt.title('Histograma de Calificaciones - Grupo 2')
plt.xlabel('Calificación')
plt.ylabel('Frecuencia')
plt.grid(True)
plt.show()
```

```python
print('GRUPO 3')
print(df3.head())
print(df3.describe())
plt.figure(figsize=(6,4))
plt.hist(df3['Calificacion'], bins=10)
plt.title('Histograma de Calificaciones - Grupo 3')
plt.xlabel('Calificación')
plt.ylabel('Frecuencia')
plt.grid(True)
plt.show()
```

```python
from sklearn.metrics import r2_score

nombre = 'Grupo 1'
df = df1
x = df[['HorasEstudio']]
y = df['Calificacion']

modelo = LinearRegression()
modelo.fit(x, y)

pendiente = modelo.coef_[0]
intercepto = modelo.intercept_

y_pred = modelo.predict(x)

correlacion, p_valor = pearsonr(df['HorasEstudio'], df['Calificacion'])

r2 = r2_score(y, y_pred)

print(f'===== {nombre} =====')
print(f'Pendiente: {pendiente:.4f}')
print(f'Intercepto: {intercepto:.4f}')
print(f'Ecuación: y = {pendiente:.4f}x + {intercepto:.4f}')
print(f'Correlación de Pearson: {correlacion:.4f}')
print(f'Valor p: {p_valor:.10f}')
print(f'R²: {r2:.4f}')

plt.figure(figsize=(7,5))
plt.scatter(x, y, label='Datos reales')
plt.plot(x, y_pred, linewidth=2, label='Línea de regresión')

plt.title(f'Regresión Lineal - {nombre}')
plt.xlabel('Horas de estudio')
plt.ylabel('Calificación')
plt.legend()
plt.grid(True)
plt.show()
```
